### LOC - Localización por index

```python
### LOC: Localiza las filas por indice, etiquetas , dado que las columnas no tienen indice se debe usar su etiqueta o nombre
```

```python
##m Muestra el rango de los indices
# ventas_df.loc [0: 4]
## Muestra el rango del indicio filtrando columnas
ventas_df.loc[0: 4, ["vendedor", "producto"]]
```

```python
# DEL INDEX 0 HASTA EL INDEX 1000
ventas_df.loc[:1000, ["vendedor", "producto"]]
```

```python
# DEL INDEX 1000 HASTA EL FINAL
ventas_df.loc[1000:, ["vendedor", "producto"]]
```

```python
# DEVUELVE TODOS LOS INDEX
ventas_df.loc[:, ["vendedor", "producto"]]
```

### ILOC - SLICING SIMILAR A PYTHON NATIVO (EX. LISTAS)

```python
### TRABAJA COMO INDEXACION COMO PYTHON CON RANGOS DE FILAS Y COLUMNAS NUMERICAS QUE EMPIEZASN DE 0 A UN RANGO -1
ventas_df.iloc[0: 4]
```

```python
ventas_df.iloc[0: 4, 0: 3]
```
