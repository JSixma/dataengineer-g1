TIPO DE VARIABLES

```python
print(type(cantidad))
print(type(precio))
print(type(descuento))
```

CASTEO DE VARIABLES "convertir un tipo de variable a otro"

```python
cantidad = "3"
print(type(cantidad))

cantidad = int(cantidad)
print(type(cantidad))
```

```python
activo = -1000
print(activo)

activo = bool(activo)
print(activo)
```
